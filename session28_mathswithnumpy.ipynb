Calculating variance of a list

```python
l=[10,20,30,40,50,60,70,80]
s=0
for i in l:
    s+=i
mean=s/len(l)
print(mean)
sub=[i-mean for i in l]
square=[i*i for i in sub]
variance=sum(square)/len(l)
variance
```

```python
np.sum(np.square((np.array(l)-np.mean(np.array(l)))))/np.array(l).size
```

```python
arr=np.array(l)
np.var(arr)
```

```python
print(np.var(arr))
```

```python
np.sqrt(np.var(arr))
```
