**Numpy**
- Numerical python or Numer python
- We can represent numbers in many formats
   - list or tuple
   - numpy
   - tensors
-  For AI model development we need vector representation of data
-  Vector representation can be done using numpy and tensors
-  we cannot perform maths and physics operations using list or tuple
-  we need numpy or tensors
-  For cpu numpy is enough
-  If we work for big data then prefer tensors

```python
import numpy as np
np.__version__
```

```python
l=[10,20,30,40]
arr1=np.array(l)
arr1
```

```python
type(arr1)
```

```python
arr2=np.array([100,200,300,400])
arr2
```

```python
#Element wise operations
print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
```

```python
import time
starttime=time.time()
l1=range(1,10000000)
l2=range(1,10000000)
[i*j for i,j in zip(l1,l2)]
endtime=time.time()
print(endtime-starttime)
```
