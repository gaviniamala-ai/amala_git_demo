**consine similarity**

```python
arr1=np.array([1,1])
arr2=np.array([1,2])
from numpy.linalg import norm
nr=np.sum(arr1*arr2)
dr=(norm(arr1)*norm(arr2))
dotproduct=nr/dr
dotproduct
```

```python
arr1=np.array([10,20])
arr2=np.array([20,40])
from numpy.linalg import norm
nr=np.sum(arr1*arr2)
dr=(norm(arr1)*norm(arr2))
dotproduct=nr/dr
dotproduct
```

```python
from sentence_transformers import SentenceTransformer
```

```python
model=SentenceTransformer('all-MiniLM-L6-V2')
```

```python
model
```

```python
sentences=['Modi is the PM of India',
           'Virat Kohli is a cricketer',
           'Messi retired from football']
question=['who is pm of India']
```

```python
model.encode('king')
```

```python
sent_vec=model.encode(sentences)
```

```python
qv=model.encode(question)
```

```python
v1=sent_vec[0]
v2=sent_vec[1]
v3=sent_vec[2]
```

```python
import numpy as np
from numpy.linalg import norm
cs=[np.sum(v*qv)/(norm(v)*norm(qv)) for v in sent_vec]
cs
```

```python
cos_sim_result={i:j for i,j in zip(sentences,cs)}
cos_sim_result
```

```python
related_sent=sorted(cos_sim_result,key=lambda x:x[1],reverse=True)[0]
print(related_sent)
cos_sim_result[related_sent]
```
