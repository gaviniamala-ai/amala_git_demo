```python
import numpy as np
ya=np.array([2,5,7,8])
x=np.array([1,2,3,4])
```

```python
b0=0
b1=0
yp=b0+b1*x
n=ya.size
alpha=0.01
minvalue=costfunction=np.sum(ya-yp)/n
b0final=b1final=0
for i in range(1000):
    print(f"iteration {i+1}",end=" ")
    print(f"b0={round(b0,4)}\tb1={round(b1,2)}",end=" ")
    yp=b0+b1*x
    costfunction=np.sum((ya-yp)**2)/n
    if abs(minvalue)>abs(costfunction):
        minvalue=costfunction
        b0final=b0
        b1final=b1
    print(f"error = {round(costfunction,2)}")
    b0+=alpha*2/n*np.sum(ya-yp)
    b1+=alpha*2/n*np.sum((ya-yp)*x)

print(f"final b0={round(b0final,4)}\n final b1={round(b1final,4)}")    
    
```
